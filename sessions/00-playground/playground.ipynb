```python
class ExpenseTracker:
  def __init__(self, budget: int):
    self.budget = budget

  def add(self, desc:str, amount:int) -> None:
    if amount <=0:
      raise ValueError("Pls input only positive amount.")
    self.budget = self.budget - amount

  def remaining(self) -> int:
    return self.budget



trip = ExpenseTracker(budget=100)
trip.add("train", 30)



try:
  trip.add("hostel", -45)
  trip.remaining()
except ValueError as e:
  print(type(e))
  print(e)
  print(help(e))


  

```
