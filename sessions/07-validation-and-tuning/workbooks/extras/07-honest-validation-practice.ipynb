# Honest validation: a small experiment

Session 07. Six synthetic listings, not measured course benchmark data. Predict each host’s price from their training mean; fall back to the overall training mean for a new host.

Run this notebook with Python 3. It needs no third-party packages.

```python
rows = [('A1', 'Ada', 100), ('A2', 'Ada', 110), ('B1', 'Ben', 180), ('B2', 'Ben', 190), ('C1', 'Cleo', 260), ('C2', 'Cleo', 270)]

def evaluate(test_ids):
    train = [r for r in rows if r[0] not in test_ids]
    valid = [r for r in rows if r[0] in test_ids]
    errors = []
    for identifier, host, price in valid:
        known = [r[2] for r in train if r[1] == host]
        values = known or [r[2] for r in train]
        prediction = sum(values) / len(values)
        errors.append(abs(price - prediction))
        print(identifier, 'actual:', price, 'prediction:', prediction)
    return sum(errors) / len(errors)

print('Random demonstration MAE:', evaluate({'A2', 'C2'}))
print('Unseen-host demonstration MAE:', evaluate({'C1', 'C2'}))
```
