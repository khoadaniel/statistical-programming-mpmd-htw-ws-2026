# Data careers, the course and the challenge

Data work combines a question, evidence and a decision. Job titles are not precise definitions: the same title can mean different work in different organisations, and one person may cover several functions. The tasks and outputs are more useful than the title when reading a job description. This notebook first describes the roles and what Berlin employers ask for, then looks at how AI coding agents change data work and learning, explains how the course is organised and assessed, and ends with the challenge that all teams work on until the final presentations: predicting the customs heading of a product from its description.

## Common roles

| Role | Main responsibility | Typical outputs |
|---|---|---|
| Data analyst | Describe what happened and explain patterns for a decision | Queries, charts, dashboards, reports and statistical comparisons |
| Data scientist | Build and evaluate statistical or predictive methods | Experiments, forecasts, models and uncertainty estimates |
| Data engineer | Make data available and trustworthy | Data pipelines, tables, quality checks and documentation |
| ML engineer | Put models into reliable use | Training or inference pipelines, services, monitoring and tests |
| AI engineer | Build and evaluate applications using existing AI models | Search or assistant systems, retrieval, evaluations and operational controls |

These are overlapping descriptions, not a ranking. An analyst may build a model; a data scientist may maintain a pipeline; an ML engineer may write SQL and analyse failures. Some organisations also use **analytics engineer** for the work of modelling raw data into trusted tables for analysts. The role boundaries depend on the team and the product.

Consider a Berlin accommodation platform. A data engineer combines listing and booking records and checks whether identifiers and timestamps are valid. An analyst reports how the number and type of listings change by district. A data scientist estimates what factors are associated with nightly price and evaluates a price model. An ML engineer makes a tested prediction service available to the product and monitors its behaviour. An AI engineer might build a search assistant that explains listing policies using a language model and evaluates whether its answers are supported by the source documents. The same data and question can therefore involve several kinds of work.

The figure places the five roles along the path from raw data to a product that people use. It is a simplification: each column is a kind of work, and a real job usually covers several of them. The next section checks it against what Berlin employers actually write in their advertisements.

![Five data roles along the path from data to product: data engineers collect, store and model data; analysts analyse and report; data scientists build and evaluate models; ML engineers deploy and operate them; AI engineers build AI applications](../figures/role_map.png)

## Skills shared across roles

**Technical skills** include Python, SQL, data formats, version control and knowledge of the tools needed for the task. **Method skills** include defining a population, checking data quality, choosing an appropriate comparison, evaluating a model and describing uncertainty. **Domain skills** mean understanding what a field or result means in the real process. **Communication skills** mean explaining the decision and its limits to someone who will use it.

A strong result is more than code that runs. For example, a chart of median nightly prices by district needs a definition of “listing,” a time period, a rule for missing prices and a note that asking price is not actual booking revenue. A prediction needs an evaluation set that represents the future use. An application needs a way to detect failures after it is deployed.

A useful portfolio project makes those choices visible. Its repository should explain the question, provenance and limitations of the data, the analysis or model, the checks used, and how another person can reproduce the result. An entry-level applicant can demonstrate these practices through a course project; particular degree, language and experience requirements vary by employer and location.

These numbers describe one board on one day and one survey; they show tendencies, not the whole labour market.

## Reading a job description

A job title alone does not tell you the work. Separate the advertisement into its **problem**, **deliverables**, **methods**, **tools**, **collaborators** and **requirements**:

| Heading | Question to ask | Example phrase |
|---|---|---|
| Problem | Which business question or product does the team work on? | "reduce failed deliveries" |
| Deliverables | What does the person hand over? | "dashboards", "a forecasting service" |
| Methods | Which kinds of analysis or modelling? | "A/B tests", "time-series models" |
| Tools | Which languages, libraries, platforms? | "SQL, Python, dbt, AWS" |
| Collaborators | With whom does the person work? | "product managers, data engineers" |
| Requirements | What must an applicant bring, and what is optional? | "required: 2 years SQL"; "nice to have: German" |

"Build dashboards for operations" describes different work from "deploy low-latency model inference", even if both postings say "data scientist". A requested tool is evidence of the environment, not the whole job. A long list may combine essential and optional skills, and "nice to have" is not "required".

In the exercises of this block (Part 1 of the [exercise notebook](session-01-exercises.ipynb)), groups read five real Berlin advertisements per role, which the lecturer hands out, and fill one row of this table per advertisement. Five advertisements are a reading exercise, not a labour-market survey: they cannot establish how common a skill is across the city. The heat map above, with about 40–60 ads per role, is the better basis for that question, and it has its own limits.

## Agentic engineering: how AI coding agents change data work and learning

A quarter of the Berlin ads above ask for "agentic AI", and the way code is written has changed quickly. This section explains what an AI coding agent is, what the evidence says about its effect on work, and why it changes how this course teaches.

**From vibe coding to agentic engineering.** In February 2025 Andrej Karpathy named a new habit: "There's a new kind of coding I call 'vibe coding', where you fully give in to the vibes, embrace exponentials, and forget that the code even exists." He did not read the generated code any more and said it was "not too bad for throwaway weekend projects" ([Karpathy, 2 February 2025](https://x.com/karpathy/status/1886192184808149383)). One year later he proposed a different name for professional use, *agentic engineering*: "the new default is that you are not writing the code directly 99% of the time", but orchestrating agents that do and "acting as oversight" ([Karpathy, 4 February 2026](https://x.com/karpathy/status/2019137879310836075)).

An **agent** is a language model that runs tools in a loop: it reads files, writes code, runs it, reads the error or the test result and tries again ([Willison, *What is agentic engineering?*, 2026](https://simonwillison.net/guides/agentic-engineering-patterns/what-is-agentic-engineering/)). The human part does not disappear; it moves. Willison: "The craft has always been figuring out what code to write." Anthropic's own guide to its coding agent puts the rule plainly: "Always provide verification (tests, scripts, screenshots). If you can't verify it, don't ship it." ([Anthropic, *Best practices for Claude Code*](https://www.anthropic.com/engineering/claude-code-best-practices)).

```mermaid
flowchart LR
  H1["You: frame the problem,<br/>write the specification<br/>and the checks"] --> A["Agent: plans, writes<br/>and runs code"]
  A --> T{"Tests and checks<br/>pass?"}
  T -- no --> A
  T -- yes --> H2["You: read the result,<br/>question it, decide"]
  H2 -- not convinced --> H1
  H2 -- convinced --> S["Ship or report"]
```

**What the evidence says about productivity.** The numbers are mixed, and measuring the effect is harder than it sounds:

| Study | What was measured | Result |
|---|---|---|
| [METR, July 2025](https://metr.org/blog/2025-07-10-early-2025-ai-experienced-os-dev-study/) | Randomised trial: 16 experienced open-source developers, 246 real tasks, AI allowed or not | 19 % *slower* with AI, although they believed they were 20 % faster |
| [METR, February 2026](https://metr.org/blog/2026-02-24-uplift-update/) | Follow-up with 57 developers | Point estimates of 18 % and 4 % speed-up, with confidence intervals that include no effect; METR calls it "an unreliable signal", partly because 30 to 50 % of developers would not do some tasks without AI |
| [Cui et al., *Management Science*, 2026](https://doi.org/10.1287/mnsc.2025.00535) | Three field experiments, 4,867 developers, a code-completion assistant | 26 % more completed tasks; larger gains for less experienced developers |
| [Stack Overflow survey, 2025](https://survey.stackoverflow.co/2025/ai) | 49,000+ developers | 84 % use or plan to use AI tools; 46 % distrust their accuracy, 33 % trust it; the top frustration (66 %) is "AI solutions that are almost right, but not quite" |
| [Google DORA, 2025](https://cloud.google.com/blog/products/ai-machine-learning/announcing-the-2025-dora-report) | Survey of nearly 5,000 technology professionals | 90 % use AI at work; "AI doesn't fix a team; it amplifies what's already there" |

Two lessons follow. Self-reported speed is not measured speed. And "almost right" is the normal failure: code that runs and looks plausible but is wrong in a detail.

**What it means for jobs.** Using US payroll data up to June 2026, Brynjolfsson, Chandar and Chen find no economy-wide displacement, but employment of 22- to 25-year-olds in AI-exposed occupations is 19 % below where it would have been, mainly because fewer young people are hired; the authors call this descriptive, not causal ([*Canaries in the Coal Mine?*, revised August 2026](https://digitaleconomy.stanford.edu/publications/canaries-in-the-coal-mine/)). The Yale Budget Lab found no discernible disruption of the US labour market as a whole ([October 2025](https://budgetlab.yale.edu/research/evaluating-impact-ai-labor-market-current-state-affairs)). This matches the Berlin picture above: entry is the hard part.

**What stays human in data work.** When Anthropic surveyed about 9,700 users of its assistant in 2026, the most common reason they gave for not delegating more was that "AI lacks the judgment, contextual awareness, and situational reasoning that their work requires" ([Anthropic Economic Index, June 2026](https://www.anthropic.com/research/economic-index-june-2026-report)). Its own engineers delegate above all "tasks that are easily verifiable" ([Anthropic, December 2025](https://www.anthropic.com/research/how-ai-is-transforming-work-at-anthropic)). For data work this is the core risk: an analysis can run without errors and still answer the wrong question, use the wrong population or leak the answer into the features. A 2026 preprint finds that language-model agents "frequently make subtle inferential errors that lead to incorrect conclusions despite correctly executed analyses" in hypothesis testing ([Miao et al., arXiv:2608.07437](https://arxiv.org/abs/2608.07437)). Framing the question, understanding the data, validating the result and explaining it to a stakeholder are the parts the roles above are paid for.

**English as a programming language, with a caveat.** "The hottest new programming language is English", wrote Karpathy in 2023 ([24 January 2023](https://x.com/karpathy/status/1617979122625712128)), and in 2025 he described language models as "a new kind of computer, and you program them in English" ([*Software Is Changing (Again)*, YC AI Startup School, June 2025](https://www.youtube.com/watch?v=LCEmiRjPEtQ)). Natural language lowers the barrier to building software. Edsger Dijkstra's old warning still holds, though: natural language makes it easy to write "statements the nonsense of which is not obvious" ([EWD 667, about 1978](https://www.cs.utexas.edu/~EWD/transcriptions/EWD06xx/EWD667.html)). The precision does not disappear; it moves into the specification, the tests and your ability to read what the agent wrote.

**How this course teaches because of it.** You do not need to memorise syntax: an assistant or the documentation supplies it in seconds. You do need to understand code well enough to read it, test it and notice when it is wrong, and that understanding does not come from delegating:

- In a randomised trial with 52 mostly junior engineers learning a new library, the group that used AI scored 50 % on a follow-up quiz against 67 % for the group that coded by hand, with the largest gap in debugging, and finished only about two minutes faster ([Shen and Tamkin, Anthropic, January 2026](https://www.anthropic.com/research/AI-assistance-coding-skills)). Participants who used AI to ask for explanations, not only for code, kept most of their learning.
- In a field experiment with nearly 1,000 high-school students, unrestricted access to GPT-4 raised grades during practice, but grades fell by 17 % once the access was removed; a tutor version with guardrails avoided most of the loss ([Bastani et al., *PNAS*, 2025](https://doi.org/10.1073/pnas.2422633122)).
- Computing educators conclude that teaching should put "less emphasis on writing code and more emphasis on reading code" and more on problem decomposition ([Prather et al., ITiCSE working group, 2023](https://doi.org/10.1145/3623762.3633499); [Denny et al., *Communications of the ACM*, 2024](https://doi.org/10.1145/3624720)).

So the exercises of this course ask you to predict what code does, to read and fix code, to write the check before the solution and to defend a result in the debrief. Use AI assistants as a tutor that explains, not as a ghostwriter: in the final project you may use them, but every member must be able to explain every line the team submits.

## Course organisation and assessment

The course has 18 sessions on 14 teaching days between 22 October 2026 and 12 February 2027; the dates are in the [course README](../../../README.md#sessions). It moves through the same work as the roles above rather than teaching isolated packages:

| Sessions | Part | Roles it prepares most |
|---|---|---|
| 1–2 | Python for analysis and for software: classes, tests, Git, continuous integration | all |
| 3–4 | Data: SQL with SQLite, data quality and preparation | all, data engineers |
| 5 | Exploration and statistics | analysts |
| 6–12 | Machine learning: regression, validation, classification, features, trees, clustering, forecasting | data scientists |
| 13–16 | Text, language models, web APIs, deployment and monitoring | data scientists, ML and AI engineers |
| 17–18 | Final presentations | all |

Sessions 1–12 use the Berlin Airbnb listings as their running example (and a telecom churn table for classification); Sessions 13–16 use the customs decisions of the final project.

**Every 90-minute block has the same rhythm:** 45 minutes of theory, in which the lecturer works through a lecture notebook like this one; 30 minutes of exercises, which you solve in pairs or teams from the session's exercise notebook; and a 15-minute debrief, in which two or three solutions, one of them wrong, are compared. The exercises are not graded: they are where you practise, and a wrong solution in the debrief is useful to everyone.

**Assessment.** The module is assessed by one component: the **final project**, presented by each team in Session 17 or 18 (100 % of the grade). Teams have three members (one team four). The grade combines five criteria assessed per team and one assessed per member, as set out in the [final project brief](../../17-18-final-presentations/final-project.md#presentation-and-assessment):

| Criterion | Weight |
|---|---|
| Problem definition | 10 % |
| Data | 20 % |
| Model | 25 % |
| Reliability of the results | 15 % |
| Delivery (a tested service, monitoring plan, model card) | 10 % |
| Collaboration and presentation (per member) | 20 % |

AI tools are allowed and their use is documented with the HTW declaration; every member must be able to explain any part of the code the team submits. The brief is the authoritative source for milestones, rules and criteria.

## The final project challenge

Every product that enters or leaves the EU needs a tariff code, which determines the customs duty and the trade statistics. When a trader is unsure about the code, the customs authority of a member state can issue a **Binding Tariff Information** (BTI) decision: it describes the goods, states the code and explains why. The European Commission publishes these decisions in the **EBTI** database.

**The task.** Given the description of goods, which four-digit **heading** of the Harmonized System will customs assign? All teams work on this one task, and a class leaderboard compares their predictions on decisions the models have not seen.

| Field of one decision | Example | Role in the project |
|---|---|---|
| Description of goods | *A ladies casual shoe with leather uppers and plastic outer soles. Not covering the ankle. Lace up …* (shortened) | **input** |
| Issuing country, language, date | United Kingdom, English, November 2017 | **input** |
| Heading | **6403**: footwear with outer soles of rubber, plastics, leather or composition leather and uppers of leather | **target** |
| Keywords, justification, CN code | written by customs together with the decision | **not allowed as input** |

Small differences in the goods change the heading: the same shoe with uppers of textile belongs to heading 6404, with uppers of plastic to 6402. A model has to learn such distinctions from the text.

**The data**, counted in the prepared files of the course (`case-study/prepare_data.py`):

- 309,529 decisions from 2017–2023 for training, and 113,188 decisions from 2024–2026 without headings as the test set: 2024 for the public leaderboard, 2025–2026 for the private one, shown in Session 18.
- 1,114 different headings in the training data. The most frequent one covers only 4 % of the decisions, and a quarter of the headings have fewer than ten examples.
- Descriptions in the language of the issuing country: 57 % German, 16 % French, 5 % English and the rest in about 20 other EU languages.

![Share of training decisions by language: German 57 %, French 16 %, English 5 %, Dutch 5 %, and smaller shares for the other languages](../../../proposal/assets/final-project/languages.png)

**How the project runs.** Teams are formed today. They set up a team repository with tests and continuous integration (Session 2), load and check the decisions with SQL (Sessions 3–4), explore them and build baselines and a validation plan (Sessions 5–9), and build text models in Sessions 13–16, with three leaderboard rounds. Accuracy is the main metric; the leaderboard score itself is not converted into marks, because the grade rewards how a team frames, builds, validates and explains its solution.

**Two rules from the start.** The test decisions can be looked up in the public EBTI database, and doing so is not allowed. Columns that customs write with or after the decision (justification, keywords, CN code) must not be model inputs: a model must work with what exists when a new request arrives. Both rules are explained in the [final project brief](../../17-18-final-presentations/final-project.md); Session 13 shows what happens to a model that ignores the second one.

The datasets are not stored in the repository: each student downloads them with the course scripts (Inside Airbnb now, EBTI before Session 3). Their sources and licences are listed in the [session README](../README.md#sources-and-licences).
