## Common roles

Seven titles are among the most common in Berlin data job advertisements today. They are ordered here along the path that data travels, from the systems where it is created to the products and decisions that use it:

| Role | Main responsibility | Typical outputs | Who uses the outputs |
|---|---|---|---|
| Data platform engineer | Build and run the shared data platform: storage, compute, access, automation | A reliable, secure, affordable platform; templates and tools for the other roles | All other data roles |
| Data engineer | Bring data from many sources into one central place, reliably and on time | Ingestion pipelines, raw and cleaned tables, data-quality checks | Analytics engineers, analysts, data scientists |
| Analytics engineer | Turn raw tables into trusted, documented tables that answer business questions | Transformation pipelines (often in SQL with dbt), tested data models, metric definitions | Analysts, data scientists, business users |
| Data analyst | Describe what happened and explain patterns for a decision | Queries, charts, dashboards, reports and statistical comparisons | Managers and teams who decide |
| Data scientist | Build and evaluate statistical or predictive methods | Experiments, forecasts, models and uncertainty estimates | Product teams, analysts, ML engineers |
| ML engineer | Put models into reliable use | Training and prediction pipelines, services, monitoring and tests | Products and their users |
| AI engineer | Build and evaluate applications on top of existing AI models | Search or assistant systems, retrieval, evaluations and guardrails | Products and their users |

These are overlapping descriptions, not a ranking. An analyst may build a model; a data scientist may maintain a pipeline; an ML engineer may write SQL and analyse failures. In a small company one person may do the work of the first three rows. The next subsections start with the data engineer, whose work is the least visible from the outside, then the data platform and the engineers who run it, then the analytics engineer, and finally the other four roles.

### Data engineers: getting data into one place

A company's data is created in many places: the database behind its app or website, a payment provider, tracking of clicks and page views, a customer-relationship tool, files from partners, spreadsheets. Each system stores data in its own format and for its own purpose. Nobody can answer "how much revenue did returning customers bring last month?" while the data is spread over ten systems.

**Data engineers build the pipelines that bring this data together in one central place**, usually a *data warehouse* (a database built for analysing large amounts of data) or a *lakehouse* (data kept as files in cheap cloud storage that can still be queried like warehouse tables). A **pipeline** is a program that runs on a schedule or continuously, takes data from a source, checks it and writes it into the central store. The steps have names that appear in every job advertisement:

- **Ingestion**: copying data from a source, either in regular batches (every night, every hour) or as a continuous *stream* of events.
- **ETL or ELT**: *extract, transform, load*. In the older order (ETL) the data is cleaned before it is loaded; in the order common today (ELT) the raw data is loaded first and transformed inside the warehouse, which is where analytics engineers take over.
- **Orchestration**: a scheduler (such as Airflow or Dagster) that runs the pipelines in the right order, retries failures and raises an alarm when data is late.

A data engineer's work is judged by reliability: is today's data complete, on time and correct? Joe Reis and Matt Housley describe the work as a lifecycle with five stages, "Generate, Store, Ingest, Transform, Serve" ([Reis, *The Data Engineering Lifecycle and Undercurrents, 4 Years Later*, 27 July 2026](https://joereis.substack.com/p/the-data-engineering-lifecycle-and); first set out in their book *Fundamentals of Data Engineering*, O'Reilly, 2022).

### What a data platform is

**A data platform is the shared set of systems and tools on which all data work in a company runs.** It is not one product you buy but a combination: where data is stored, where computations run, how pipelines are scheduled, who may see which data, and how problems are detected.

An everyday analogy: a well-equipped workshop. The people who run the workshop do not build the furniture. They provide the benches, power tools, safety rules, a stockroom with labelled material and a booking system, so that every carpenter can start working without first wiring the electricity. In a company, the "carpenters" are the analysts, analytics engineers, data scientists and ML engineers.

| Part of the platform | What it does | Examples of tools |
|---|---|---|
| Storage | keeps all data in one place, raw and cleaned | Snowflake, BigQuery, Databricks, files in cloud storage |
| Compute | runs queries and processing jobs | the warehouse's engine, Spark |
| Ingestion | copies data in from the sources | Fivetran, Airbyte, Kafka |
| Orchestration | runs pipelines in order and on time, retries failures | Airflow, Dagster |
| Transformation | turns raw tables into clean, tested ones | dbt (SQL with tests and documentation) |
| Catalogue and access control | lets people find data and decides who may see what (personal data, GDPR) | Unity Catalog, DataHub |
| Quality and observability | raises an alarm when data is late, missing or wrong | dbt tests, Great Expectations, data-observability tools |
| Containers | runs each program in a container (see the next subsection) | Docker, Kubernetes |
| Infrastructure as code and CI/CD | sets up the platform from code; *continuous integration and delivery* (CI/CD) tests and releases every change automatically (see the next subsection) | Terraform, GitHub Actions, GitLab CI |

#### Self-service

The point of a platform is **self-service**: an analyst can find a table, check what a column means, write a query and publish a dashboard without asking an engineer to set anything up first. Without a platform, every new question becomes a ticket for the data engineering team, which then becomes the bottleneck of the whole company. Zhamak Dehghani, who described the *data mesh* approach, calls the platform "a high-level abstraction of infrastructure that removes complexity and friction of provisioning", so that teams "can create and consume data products autonomously" ([Dehghani, *Data Mesh Principles and Logical Architecture*, martinfowler.com, 3 December 2020](https://martinfowler.com/articles/data-mesh-principles.html)). Data and platform engineers therefore do not only deliver tables; by building the platform they **enable the other data roles to work on their own**.

### Platform engineering, infrastructure and DataOps

Someone has to build and run the platform itself. In large companies this is a separate role, the **data platform engineer** (also *data infrastructure engineer*); in smaller ones it is part of the data engineer's job. This work is about **infrastructure**, the machines and services underneath the data:

- setting up cloud accounts, networks, storage and compute, and keeping their cost under control;
- running services such as the orchestrator or a streaming system, often in *containers* (a program packed with everything it needs, so that it runs the same everywhere) on *Kubernetes* (a system that runs many containers on a group of machines and restarts them when they fail);
- describing all of it as code (*infrastructure as code*, for example with Terraform, where servers, storage and networks are written down in text files and created from them), so that it can be reviewed, versioned and rebuilt;
- security and access rights, backups, and being *on call* (reachable outside working hours) when something breaks at night.

**DataOps** is the way of working that keeps this reliable. It is not a tool or usually a job title, but a set of practices borrowed from software operations (DevOps) and applied to data: version control for all pipeline code, automated tests of code *and* data, continuous integration and delivery, monitoring, and small, frequent, reversible changes. The [DataOps Manifesto](https://dataopsmanifesto.org/) states principles such as "Analytics is code", "Make it reproducible" and "Quality is paramount". Reis and Housley list DataOps among the "undercurrents running underneath all five stages" of the lifecycle, together with security, data management, data architecture, orchestration and software engineering ([Reis, 2026](https://joereis.substack.com/p/the-data-engineering-lifecycle-and)).

How much time this takes is visible in practice. In the *2026 State of Data Engineering Survey* (1,101 respondents, the largest group of them data engineers, February 2026), each person named up to two activities that take most of their time: a quarter named infrastructure and platform work (25.1 %) and about as many "fighting fires" (26.2 %); the most common answer, data modelling and transformation (55.4 %), is the work of the next subsection ([Practical Data Community, 2026](https://joereis.github.io/practical_data_data_eng_survey/)). The survey itself warns that its respondents are "engaged, senior-leaning practitioners", not a random sample.

The course meets this work in several places: tests, Git and continuous integration (Session 2), loading data into a database with SQL (Session 3), data-quality checks written as automated tests (Session 4), and containers, continuous delivery and monitoring (Session 16).

### Analytics engineers: from raw tables to trusted data models

**Analytics engineer** is a young title. dbt Labs, the company behind the transformation tool dbt, write that "starting in 2018" they and the Locally Optimistic community "started calling this role the analytics engineer", because the work "is neither data engineering, nor analysis. It's somewhere in the middle" ([dbt Labs, *What is analytics engineering?*](https://www.getdbt.com/analytics-engineering/), first published 2019). Today it is a common title: 153 of the 1,101 respondents of the 2026 data engineering survey above (13.9 %) call themselves analytics engineers, and the Berlin board listed 17 hands-on analytics engineer ads on 17 October 2026.

Analytics engineers work **inside the warehouse**, after the data has arrived. They write **transformation pipelines**, usually in SQL with dbt, that turn raw tables into clean, tested and documented *data models*: one table of customers, one of orders, one agreed definition of "active user" or "revenue". (A *data model* here is a designed table with agreed meanings, not a statistical model.) In dbt's words, an analytics engineer "spends their time transforming, testing, deploying, and documenting data" and provides "clean data sets to end users". They bring software practice into analysis: every model is code in Git, reviewed in a pull request (a proposed change that a colleague reviews; Session 2) and tested automatically before it reaches a dashboard.

The difference from a data engineer lies in **where** they work and **what they do not do**:

| | Data engineer | Analytics engineer |
|---|---|---|
| Starts from | the source systems (apps, APIs, event streams) | raw tables already in the warehouse |
| Main work | ingestion pipelines, often also the platform and its infrastructure | transformation pipelines and data models |
| Main tools | Python, SQL, orchestration, streaming, cloud, sometimes Terraform and Kubernetes | SQL, dbt, Git, the warehouse, a BI (business intelligence) tool such as Looker, Tableau or Power BI |
| Infrastructure and DataOps | part of the job in many teams, especially small ones | rarely: they use the platform rather than run it |
| Works most with | platform and software engineers | analysts and business teams |

**Predict.** The Berlin board listed 17 hands-on analytics engineer ads on 17 October 2026. How many of them name dbt, and how many name Kubernetes, Terraform or infrastructure as code? Write down two numbers before you read on.

The Berlin advertisements show this split (see the heat map below): **16 of 17** analytics engineer ads name dbt, but only **5 of 17** name Kubernetes, Terraform or infrastructure as code, four of them as a plus (one company's ad in two seniorities and two languages); for data platform engineers it is 3 of 15 and 11 of 15. The heat map's wider group "Infrastructure, DataOps" (which also counts Docker, observability and DevOps) reaches 7 of 17 analytics engineer ads (41 %), five of them as a plus. Data engineers sit between the two: 59 % of their ads name an infrastructure keyword and 26 of 66 name dbt, so in many companies the data engineer also does part of the platform and part of the modelling work. Analytics engineers do use version control and CI/CD for their models, which is software practice, not running infrastructure.

### Analysts, data scientists, ML and AI engineers

The other four roles use the data that the first three make available:

- **Data analysts** answer questions with the trusted tables: queries, charts, dashboards and statistical comparisons that support a decision. With good data models from analytics engineers, an analyst spends time on the question rather than on cleaning.
- **Data scientists** build and evaluate statistical and predictive methods: experiments (A/B tests), forecasts and models, always with an estimate of how uncertain they are.
- **ML engineers** turn a model into a reliable part of a product: training and prediction pipelines, a service with tests, monitoring of its quality over time. Their work overlaps with platform engineering, because a model needs infrastructure to run on; the same holds for AI engineers, about half of whose Berlin ads name Docker, Kubernetes or similar tools (52 %).
- **AI engineers** build applications on top of existing AI models such as large language models: search and assistant systems, retrieval of documents, and evaluations and guardrails that check whether the answers are supported by sources.

### One company, seven roles

Consider a Berlin accommodation platform, with listings, bookings, payments and reviews:

- The **data platform engineer** sets up the warehouse, the orchestrator and access rights, so that only the finance team sees payment details, and keeps the cloud bill under control.
- The **data engineer** builds pipelines that copy listings and bookings from the app's database every hour, payments from the payment provider every night, and click events as a stream, and checks that identifiers and timestamps are valid.
- The **analytics engineer** writes the dbt models that turn these raw tables into one trusted table of bookings per listing and night, with one agreed definition of "occupancy", tested and documented.
- The **data analyst** uses that table to report how the number and type of listings change by district.
- The **data scientist** estimates which factors are associated with the nightly price and evaluates a price model.
- The **ML engineer** makes a tested price-suggestion service available in the app and monitors its behaviour.
- The **AI engineer** builds a search assistant that explains listing policies with a language model and evaluates whether its answers are supported by the policy documents.

## Skills shared across roles

Four kinds of skill appear in every role:

- **Technical skills** include Python, SQL, data formats, version control and knowledge of the tools needed for the task.
- **Method skills** include defining a population, checking data quality, choosing an appropriate comparison, evaluating a model and describing uncertainty.
- **Domain skills** mean understanding what a field or result means in the real process.
- **Communication skills** mean explaining the decision and its limits to someone who will use it.

**Predict.** A chart of median nightly prices by district runs without an error. Name two decisions that could still make it wrong, then read on.

A strong result is more than code that runs. For example, a chart of median nightly prices by district needs a definition of "listing", a time period, a rule for missing prices and a note that asking price is not actual booking revenue. A prediction needs an evaluation set that represents the future use. An application needs a way to detect failures after it is deployed.

A useful portfolio project makes those choices visible. Its repository should explain the question, provenance and limitations of the data, the analysis or model, the checks used, and how another person can reproduce the result. An entry-level applicant can demonstrate these practices through a course project; particular degree, language and experience requirements vary by employer and location.

## What Berlin job ads ask for

### How the heat map was made

The heat map below is the course's own analysis, not a published statistic. We collected the advertisements ourselves from the [Data Berlin job board](https://databerlin.net/jobs), which lists data and AI jobs in Berlin, with the script [`job_ad_skills.py`](../figures/job_ad_skills.py) on **17 October 2026**:

1. **Collect.** The script loaded the board's list of open jobs (1,071 that day) and downloaded the page of every ad whose title names one of the seven roles. This is called *web scraping* or *crawling*; the board's `robots.txt` allows it, and the script waits between requests.
2. **Assign a role by title.** For example, "Senior Analytics Engineer" counts as analytics engineer and "Data Infrastructure Engineer" as data platform engineer. Titles with lead, head, principal, staff, manager, director, VP or chief were left out, so the map shows hands-on jobs. Eight ads had been taken down between loading the list and downloading the ad, and the same text posted twice was counted once: 270 ads remain, between 15 and 66 per role.
3. **Search for skills.** A skill group counts as named when one of its keywords appears in the ad text; "Data modelling, dbt", for instance, matches *dbt*, *data model*, *star schema* or *semantic layer*. The full keyword lists are at the top of the script.
4. **Count.** Each cell is the share of a role's ads that name the skill group.

Keyword matching is crude. It shows what employers **mention**, not how much of the job a skill is, and it cannot tell "required" from "nice to have". The board changes every day, so the map describes one day. The rows of data platform and analytics engineers rest on 15 and 17 ads: one ad more or less moves a cell by six or seven percentage points, and one company that posts the same job in two languages counts twice. The skill group "German" counts ads that *mention* German; an ad written in German without the word does not count.

## Reading a job description

A job title alone does not tell you the work. Separate the advertisement into its **problem**, **deliverables**, **methods**, **tools**, **collaborators** and **requirements**:

| Heading | Question to ask | Example phrase |
|---|---|---|
| Problem | Which business question or product does the team work on? | "reduce failed deliveries" |
| Deliverables | What does the person hand over? | "dashboards", "a forecasting service" |
| Methods | Which kinds of analysis or modelling? | "A/B tests", "time-series models" |
| Tools | Which languages, libraries, platforms? | "SQL, Python, dbt, Snowflake, Terraform" |
| Collaborators | With whom does the person work? | "product managers, data engineers" |
| Requirements | What must an applicant bring, and what is optional? | "required: 2 years SQL"; "nice to have: German" |

**Predict.** Two postings are both titled "data scientist". One asks you to "build dashboards for operations", the other to "deploy low-latency model inference". Is it the same job, with the same skills?

"Build dashboards for operations" describes different work from "deploy low-latency model inference", even if both postings say "data scientist". A requested tool is evidence of the environment, not the whole job. A long list may combine essential and optional skills, and "nice to have" is not "required".

## Worked example · Reading one Berlin advertisement

The six headings above, applied to one real advertisement: a *Manufacturing Data Analyst* at Almetra, from the [Data Berlin job board](https://databerlin.net/jobs/manufacturing-data-analyst-at-almetra-formerly-deltia), retrieved on 17 October 2026. It is not one of the advertisements of exercise 1.1. The quotes below are from the ad; its last third lists benefits, which say nothing about the work.

### Step 1 · Find the problem

The first sentence names it: the analyst will "translate manufacturing data into actionable insights", improving "both customer outcomes and our AI analytics platform". The team works on its customers' production lines, so the data are production and sensor data, not web clicks or sales.

**Why this step?** Why start with the problem, and not with the list of tools?

### Step 2 · List the deliverables

What the person hands over, from the part *What you'll do*: dashboards and reports on "process efficiency, bottlenecks and improvement areas"; anomalies and optimisation opportunities that "feed back into our product's analytics layer"; and "clean, well-structured code for data processing and analysis", which the ad calls "maintainable, reviewable, and production-aware".

### Step 3 · Separate methods from tools

Methods: analysing production data to "uncover performance drivers", finding anomalies, "data modelling, feature extraction and algorithmic thinking", and defining "quality metrics" with the engineers. Tools: "SQL and Python (or similar)"; manufacturing software such as MES, ERP and SCADA appears only under *Nice to have*. The tool list is short; the methods carry most of the job.

### Step 4 · Name the collaborators

Manufacturing customers, "AI and backend engineers", and the three teams the analyst connects: "Customer Success, Product and Engineering". Defining "data pipelines, features and quality metrics" with engineers is part of the job, so this analyst also does some of the work of the analytics engineer.

### Step 5 · Split what is required from what is optional

Required (the part *You are*): "3–5 years in data analytics or process optimisation", "Strong in SQL and Python", experience "with production data, sensors, or process control systems", and communicating findings "to both technical and operational stakeholders". Nice to have: a background in process engineering or lean manufacturing, manufacturing software, image processing.

**Why this step?** The ad ends with the question "Don't meet all the requirements?" and encourages applicants who "do not meet every qualification". What does that tell you about how to read the list of requirements?

*Result:* the ad's row of the table of exercise 1.1.

| Problem | Deliverables | Methods | Tools | Collaborators | Required | Nice to have |
|---|---|---|---|---|---|---|
| manufacturing data into insights for customers and the product | dashboards, reports, anomalies, production-aware code | performance drivers, anomalies, data modelling, feature extraction, quality metrics | SQL, Python | customers, AI and backend engineers, Customer Success, Product | 3–5 years, SQL and Python, production data, communication | process engineering, MES, ERP, SCADA, image processing |

The title says data analyst, but the work includes reviewed, production-aware code and pipeline work with engineers: the deliverables describe the job better than the title. ## Agentic engineering: how AI coding agents change data work and learning

Almost a third of the ads on the Berlin board (30 %, see the board's skills table in the job-ads section) name "agentic AI", and more than half name LLMs, and the way code is written has changed quickly. This section explains what an AI coding agent is, what the evidence says about its effect on work, and why it changes how this course teaches.

### From vibe coding to agentic engineering

*Vibe coding* and *agentic engineering* are informal names coined by Andrej Karpathy, not technical standards: the first means letting an AI write code you do not read, the second means directing AI agents professionally and checking their work. In February 2025 Andrej Karpathy named a new habit: "There's a new kind of coding I call 'vibe coding', where you fully give in to the vibes, embrace exponentials, and forget that the code even exists." He no longer read the diffs of the generated code and said it was "not too bad for throwaway weekend projects" ([Karpathy, 2 February 2025](https://x.com/karpathy/status/1886192184808149383)). One year later he proposed a different name for professional use, *agentic engineering*: "the new default is that you are not writing the code directly 99% of the time", but orchestrating agents that do and "acting as oversight" ([Karpathy, 4 February 2026](https://x.com/karpathy/status/2019137879310836075)).

### What an AI coding agent is

An **agent** is a language model that runs tools in a loop: it reads files, writes code, runs it, reads the error or the test result and tries again ([Willison, *What is agentic engineering?*, 2026](https://simonwillison.net/guides/agentic-engineering-patterns/what-is-agentic-engineering/)). The human part does not disappear; it moves. Willison: "The craft has always been figuring out what code to write." Anthropic's own guide to its coding agent puts the rule plainly: "Always provide verification (tests, scripts, screenshots). If you can't verify it, don't ship it." ([Anthropic, *Best practices for Claude Code*](https://code.claude.com/docs/en/best-practices), retrieved 14 October 2026).

### What the evidence says about productivity

**Predict.** In a randomised trial, 16 experienced open-source developers worked on real tasks, with and without AI. Were they faster or slower with AI, and how fast did they think they had been?

The numbers are mixed, and measuring the effect is harder than it sounds:

| Study | What was measured | Result |
|---|---|---|
| [METR, July 2025](https://metr.org/blog/2025-07-10-early-2025-ai-experienced-os-dev-study/) | Randomised trial: 16 experienced open-source developers, 246 real tasks, AI allowed or not | Tasks took 19 % *longer* with AI, although afterwards the developers estimated that AI had sped them up by 20 % |
| [METR, February 2026](https://metr.org/blog/2026-02-24-uplift-update/) | Follow-up with 57 developers | Point estimates of 18 % and 4 % speed-up, with confidence intervals that include no effect; METR calls it "an unreliable signal", partly because 30 to 50 % of developers said they did not submit some tasks because they did not want to do them without AI |
| [Cui et al., *Management Science*, 2026](https://doi.org/10.1287/mnsc.2025.00535) | Three field experiments, 4,867 developers, a code-completion assistant | 26 % more completed tasks; larger gains for less experienced developers |
| [Stack Overflow survey, 2025](https://survey.stackoverflow.co/2025/ai) | 49,000+ developers | 84 % use or plan to use AI tools; 46 % distrust their accuracy, 33 % trust it; the top frustration (66 %) is "AI solutions that are almost right, but not quite" |
| [Google DORA, 2025](https://cloud.google.com/blog/products/ai-machine-learning/announcing-the-2025-dora-report) | Survey of nearly 5,000 technology professionals | 90 % use AI at work; "AI doesn't fix a team; it amplifies what's already there" |

Two lessons follow. Self-reported speed is not measured speed. And "almost right" is the normal failure: code that runs and looks plausible but is wrong in a detail.

### What it means for jobs

Using US payroll data up to June 2026, Brynjolfsson, Chandar and Chen find no economy-wide displacement, but employment of 22- to 25-year-olds in AI-exposed occupations is 19 % below where it would have been, mainly because fewer young people are hired; the authors call this descriptive, not causal ([*Canaries in the Coal Mine?*, revised August 2026](https://digitaleconomy.stanford.edu/publications/canaries-in-the-coal-mine/)). The Yale Budget Lab found no discernible disruption of the US labour market as a whole ([October 2025](https://budgetlab.yale.edu/research/evaluating-impact-ai-labor-market-current-state-affairs)). This matches the Berlin picture above: entry is the hard part.

### What stays human in data work

When Anthropic surveyed about 9,700 users of its assistant in 2026 and asked which tasks AI would never be able to do and why, the most common responses emphasised that "AI lacks the judgment, contextual awareness, and situational reasoning that their work requires" ([Anthropic Economic Index, June 2026](https://www.anthropic.com/research/economic-index-june-2026-report)). Its own engineers delegate above all "tasks that are easily verifiable" ([Anthropic, December 2025](https://www.anthropic.com/research/how-ai-is-transforming-work-at-anthropic)). For data work this is the core risk: an analysis can run without errors and still answer the wrong question, use the wrong population or leak the answer into the features. A 2026 preprint finds that language-model agents "frequently make subtle inferential errors that lead to incorrect conclusions despite correctly executed analyses" in hypothesis testing ([Miao et al., arXiv:2608.07437](https://arxiv.org/abs/2608.07437)). Framing the question, understanding the data, validating the result and explaining it to a stakeholder are the parts the roles above are paid for.

### English as a programming language, with a caveat

"The hottest new programming language is English", wrote Karpathy in 2023 ([24 January 2023](https://x.com/karpathy/status/1617979122625712128)), and in 2025 he described language models as "a new kind of computer, and you program them in English" ([*Software Is Changing (Again)*, YC AI Startup School, June 2025](https://www.youtube.com/watch?v=LCEmiRjPEtQ)). Natural language lowers the barrier to building software. Edsger Dijkstra's old warning still holds, though: natural language makes it easy to write "statements the nonsense of which is not obvious" ([EWD 667, about 1978](https://www.cs.utexas.edu/~EWD/transcriptions/EWD06xx/EWD667.html)). The precision does not disappear; it moves into the specification, the tests and your ability to read what the agent wrote.

### How this course teaches because of it

You do not need to memorise syntax: an assistant or the documentation supplies it in seconds. You do need to understand code well enough to read it, test it and notice when it is wrong, and that understanding does not come from delegating:

- In a randomised trial with 52 mostly junior engineers learning a new library, the group that used AI scored 50 % on a follow-up quiz against 67 % for the group that coded by hand, with the largest gap in debugging, and finished only about two minutes faster, a difference that was not statistically significant ([Shen and Tamkin, Anthropic, January 2026](https://www.anthropic.com/research/AI-assistance-coding-skills)). Participants who used AI to ask for explanations or conceptual questions, not only for code, were among the higher scorers; the authors note that this link is an association, not a proven cause.
- In a field experiment with nearly 1,000 high-school students, unrestricted access to GPT-4 raised grades during practice, but once the access was removed these students scored 17 % lower than students who never had it; a tutor version with guardrails largely mitigated this loss ([Bastani et al., *PNAS*, 2025](https://doi.org/10.1073/pnas.2422633122)).
- An ITiCSE working group of computing educators suggests that instructors "might need to put less emphasis on writing code and more emphasis on reading code", and reports instructors who now teach problem decomposition explicitly ([Prather et al., 2023](https://doi.org/10.1145/3623762.3633499); see also [Denny et al., *Communications of the ACM*, 2024](https://doi.org/10.1145/3624720)).

So the exercises of this course ask you to predict what code does, to read and fix code, to write the check before the solution and to defend a result in the debrief. Use AI assistants as a tutor that explains, not as a ghostwriter: in the final project you may use them, but every member must be able to explain every line the team submits.

## The final project challenge

Every product that enters or leaves the EU needs a tariff code, which determines the customs duty and the trade statistics. When a trader is unsure about the code, the customs authority of a member state can issue a **Binding Tariff Information** (BTI) decision: it describes the goods, states the code and explains why. The European Commission publishes these decisions in the **EBTI** database.

### The task

Given the description of goods, which four-digit **heading** of the Harmonized System will customs assign? All teams work on this one task. Each team builds a small web service (a REST API) that answers with the heading, and at the end the lecturer tests every service on decisions that no team has seen.

| Field of one decision | Example | Role in the project |
|---|---|---|
| Description of goods | *Ladies Sandals – PU Weave Strap Mule Sandal. The sandal consists of a silicone rubber sole with a vamp made of straps of polyurethane weave …* (shortened) | **input** |
| Issuing country, language, date | Ireland, English, April 2024 | **input** |
| Heading | **6402**: footwear with outer soles and uppers of rubber or plastics | **target** |
| Keywords, justification, CN code | written by customs together with the decision | **not allowed as input** |

Small differences in the goods change the heading: the same sandal with straps of leather belongs to heading 6403, with straps of textile to 6404. A model has to learn such distinctions from the text.

> **Ask the class.** Read the example aloud: which parts are known when a new request arrives, and which part is the answer?

**Predict.** The decisions cover 1,009 different headings. What share of the decisions does the most frequent heading cover: about 4 %, 20 % or 50 %?

### The data

The numbers below are counted in the course's files ([case-study/README.md](../../../case-study/README.md)):

- 119,514 decisions that are valid on 15 October 2026, the day of the export: a decision binds for three years, so they started between October 2023 and October 2026.
- You receive 102,440 of them, with their headings. How to split them into training and validation data is your team's first design decision (Sessions 7 and 9).
- The other 17,074 (one fold of seven, about 14 %) are held back by the lecturer: the decisions were shuffled with a fixed seed and drawn at random, **stratified by heading and language** so that every combination of heading and language has about the same share as in your data. Because the draw is random, the years are mixed in too: decisions that started in 2026, for example, are 27 % of your data and 28 % of the held-back decisions. They are the **independent evaluation set** on which every team's service is tested after the code freeze; they are never released.
- 1,009 different headings. The most frequent one (3926, other articles of plastics) covers only 4 % of the decisions, and about two in five headings have fewer than ten examples.
![Share of decisions by language: German 61 %, French 16 %, Dutch, Czech and Spanish 4 % each, and smaller shares for the other languages](../figures/ebti_languages.png)

**What do you see?** Compare the bar for German with the bar for English: in which language will most of the descriptions your model reads be written?

The descriptions are in the language of the issuing country: 61 % German, 16 % French and the rest in 21 other EU languages. Fewer than 1 % are in English, because the United Kingdom no longer issues EU decisions.

### How the project runs

- **Teams** of three (one of four) are formed today and work together until the end.
- **The work follows the course:** a team repository with tests and continuous integration (Session 2); loading and checking the decisions with SQL (Sessions 3–4); exploration, baselines and a validation plan (Sessions 5–9); text models and the service (Sessions 13–16).
- **Reproducible, from day one.** A result counts only if it runs on another computer, not only on yours. Your repository's README must say how to rerun the training and start the service; the lecturer may follow it, and if the scripts fail, the grade drops considerably. Being reproducible is one of the most important criteria in machine learning, and what colleagues will expect of you in any data job.
- **Collaborate through GitHub:** branches, pull requests and reviews of each other's code. The lecturer reads the Git history to see how the team worked together and who did what.
- **Ask early.** The lecturer is available for questions about the task and the team work, in class or through an issue in your repository. Sharing intermediate results is not required, but recommended: it is how you get early feedback and can correct your direction in time.
- **Graded once, at the end.** Nothing is graded before the final presentation. The grade is based on your repository on GitHub, the written report and the presentation itself.
- **A score is not a grade.** Each team reports how well its final design will work; the independent evaluation then shows whether that report holds. Accuracy is the main metric, but the score is not converted into marks: the grade rewards how a team frames, builds, validates and explains its solution.

### Rules from the start

- **Train only on the data you are given.** Do not add decisions from the public EBTI database: the held-back evaluation decisions are in it too, so your model would already have seen the test (*data leakage*).
- **Predict, never look up.** The service must predict the heading from the description, not search for the decision online.
- **Use only what exists when a request arrives.** The justification, keywords and CN code are written with or after the decision, so they cannot be inputs. Session 13 shows what happens to a model that uses them.

All rules are in the [final project brief](../../17-18-final-presentations/final-project.md#rules).
