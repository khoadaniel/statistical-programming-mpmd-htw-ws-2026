## Jupyter notebooks: cells, kernel, Markdown, order of execution

A **Jupyter notebook** is a document made of **cells**. A **Markdown cell** holds text, like this one. A **code cell** holds Python code; running it (Shift+Enter, or ▶ to its left) sends the code to the **kernel**, a running Python process, and shows the result below the cell.

- The value of the **last expression** in a code cell is displayed automatically; other values are shown with `print()`.
- The number in brackets to the left of a code cell, such as `[3]`, counts the order in which cells were run.
- The kernel **remembers** every name created by a cell that has run, whatever its position in the notebook.
- **Restart** the kernel to forget everything; **Run All** runs every cell from the top.

The [Think Python introduction to Jupyter](02-jupyter-introduction.ipynb) shows the buttons and the two modes of a notebook step by step (it was written for Google Colab; in VS Code the buttons look slightly different, the ideas are the same).

````python
nightly_price = 80
nightly_price * 3          # the last expression is displayed: 240
````

**Order of execution.** The kernel does not read the notebook from top to bottom; it runs cells in the order you run them. Suppose you run the cell above, then a cell further down that sets `nightly_price = 95`, then run the cell above again: it now shows 285, although the page shows `nightly_price = 80` right above it. A notebook that only works because of such a hidden history is not reproducible.

> **Rule.** Before you share a notebook or trust its result, choose *Restart* and then *Run All*. If it fails, the error shows which cell depends on something that no longer exists.

**Markdown** is a short notation for formatted text. Double-click a Markdown cell to see it:

| You type | You get |
|---|---|
| `## Results` | a heading (one to six `#`) |
| `**median**` and `*median*` | **median** and *median* |
| `- item` | a bullet list |
| `` `price` `` | `price` as code |
| `[pandas](https://pandas.pydata.org)` | a link |

Useful keys in the notebook editor of VS Code and JupyterLab: Esc leaves the cell editor (command mode); then **A** or **B** inserts a cell above or below, **M** turns a cell into Markdown, **Y** into code, **D D** deletes it, and Enter returns to the editor.

## Variables, expressions and operators

An **expression** is code that produces a value: `80 * 3 + 20` produces 260. An **assignment** with `=` gives a value a **name**, also called a **variable**; using the name later gives back the value.

````python
nightly_price = 80
nights = 3
cleaning_fee = 20          # paid once per stay
total = nightly_price * nights + cleaning_fee
total
````

`=` is not a mathematical equation and not a spreadsheet formula. Python computes the right-hand side **once** and binds the name on the left to the result. If `nights` changes later, `total` keeps the value 260 until its line runs again.

**Names** may contain letters, digits and underscores, must not start with a digit, and are case-sensitive (`price` and `Price` are different names). Python code uses lower case with underscores (`nightly_price`), and a few words such as `if`, `for` and `None` are reserved. A good name says what the value means, including its unit if needed: `price_eur` is better than `p`.

**Operators** combine values. Python follows the usual order of operations (powers before multiplication and division, before addition and subtraction); parentheses make the intended order explicit.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `80 * 3 + 20` | `260` |
| `/` | divide (the result is always a float) | `260 / 4` | `65.0` |
| `//` | whole-number division | `17 // 7` | `2` |
| `%` | remainder of the division | `17 % 7` | `3` |
| `**` | power | `2 ** 10` | `1024` |
| `==` `!=` | equal, not equal | `total == 260` | `True` |
| `<` `<=` `>` `>=` | compare | `total <= 250` | `False` |

`//` and `%` answer questions such as "how many full weeks and how many extra nights are 17 nights?". An **augmented assignment** such as `reviews += 1` is short for `reviews = reviews + 1`.

````python
weeks = 17 // 7
extra_nights = 17 % 7
print(weeks, "weeks and", extra_nights, "nights")

reviews = 11
reviews += 1               # one more review: reviews = reviews + 1
print(reviews)
print(80 * (3 + 20))       # a different rule: the fee is charged every night
````

The last line runs without any error, but it answers a different question. Python checks that code is valid, not that it computes what you meant: check a result against a case you can compute by hand.

## Numbers, strings, booleans and None

Every value has a **type**, which decides what you can do with it. `type(value)` tells you the type.

| Type | Meaning | Examples |
|---|---|---|
| `int` | whole number, of any size | `3`, `-12`, `2 ** 100` |
| `float` | number with a decimal point | `80.0`, `52.52`, `1.5e3` |
| `str` | text (a **string** of characters) | `"Pankow"`, `'01067'`, `""` |
| `bool` | truth value | `True`, `False` |
| `NoneType` | no value | `None` |

### Numbers

An `int` is exact. A `float` is stored in binary with about 16 significant digits, so many decimal fractions are only approximated; the exercises show one surprising consequence. Round only when you **display** a result, with `round(x, 2)` or a format such as `{x:.2f}`.

````python
print(type(3), type(80.0), type(3 / 1))
price_per_guest = 160.71 / 7
print(price_per_guest)
print(round(price_per_guest, 2))
````

### Strings

A string is text between single or double quotes. Strings have a length, positions that start at **0**, and **slices** `text[start:stop]` that stop *before* `stop`. Negative positions count from the end. A string cannot be changed in place; its **methods** return new strings.

````python
title = "  Cosy flat in Pankow, 45 m² "
clean = title.strip()                  # remove spaces at both ends
print(len(clean))
print(clean[0], clean[-1], clean[:9])  # first character, last character, a slice
print(clean.lower())
print(clean.replace("Cosy", "Quiet"))
print(clean.split(","))                # a list of two strings
print("Pankow" in clean)               # membership test: True
````

An **f-string** puts values into text: write `f` before the quotes and the expression in braces. A format after a colon controls the display, for example two decimals with `:.2f`.

````python
district = "Pankow"
print(f"{nights} nights in {district} cost €{total}")
print(f"Price per guest: €{price_per_guest:.2f}")
````

**Converting between types** is always explicit: `int("80")`, `float("52.52")`, `str(80)`. Text and numbers do not mix silently:

````python
print("80" + "20")         # joins the text: 8020
print(int("80") + 20)      # converts first: 100
print(str(80) + " euros")  # 80 euros
````

`"80" + 20` stops with a `TypeError`, because `+` cannot combine text and a number. Some numbers are really **identifiers** and stay text: the postal code `"01067"` (Dresden) would lose its leading zero as an `int`, and nobody adds two postal codes.

### Booleans

A comparison produces a `bool`. `and`, `or` and `not` combine truth values. `==` asks whether two values are equal; a single `=` assigns.

````python
budget = 300
available = True
print(total <= budget)
print(available and total <= budget)
print(not available)
````

### None

`None` means "no value": a price that was not recorded, a rating of a listing that nobody has reviewed. It is not zero and not an empty string. Test for it with `is None`.

````python
rating = None
print(rating is None)      # True: the rating is missing
print(rating == 0)         # False: missing is not the same as zero
````

> **In practice.** Mixing up "missing" and "zero" changes results. If missing prices were stored as 0, the average price of a district would fall, and a model would learn that some flats are free. The data-quality session (Session 4) comes back to this.

## Lists, tuples, dictionaries and sets

Data rarely come as single values. Python has four built-in **collections**:

| Collection | Written as | Ordered | Changeable | Typical use |
|---|---|---|---|---|
| `list` | `[75, 90, 110]` | yes, by position | yes | a sequence of prices, of records |
| `tuple` | `(52.52, 13.40)` | yes, by position | no | a small fixed group: coordinates |
| `dict` | `{"district": "Pankow", "price": 75}` | by insertion | yes | a record with named fields |
| `set` | `{"Mitte", "Pankow"}` | no | yes | unique values, membership tests |

### Lists

A list keeps values in order. Indexing and slicing work as for strings, starting at 0. Unlike a string, a list can be changed in place, for example with `append`.

````python
prices = [75, 90, 110]
print(prices[0], prices[-1], prices[0:2])
prices.append(85)            # changes the list itself
print(prices, len(prices))
print(sum(prices) / len(prices))   # mean price
print(sorted(prices))              # a new, sorted list
print(110 in prices)
````

### Tuples

A tuple is like a list that cannot be changed. It suits a fixed group of values such as a pair of coordinates. A tuple can be **unpacked** into several names in one assignment.

````python
alexanderplatz = (52.5219, 13.4132)
latitude, longitude = alexanderplatz     # unpacking
print(latitude)
````

### Dictionaries

A dictionary maps **keys** to **values**. Keys are usually the names of fields, so a dictionary is a natural way to store one **record**. Looking up a key that does not exist raises a `KeyError`; `.get()` returns `None` (or a default you choose) instead.

````python
listing = {
    "district": "Pankow",
    "room_type": "Entire home/apt",
    "nightly_price": 75,
    "rating": None,
}
print(listing["district"])
listing["nightly_price"] = 80             # change a value
listing["minimum_nights"] = 2             # add a new key
print(listing.get("reviews"))             # None: the key does not exist
print(listing.get("reviews", 0))          # a default of your choice
print("rating" in listing)                # True: the key exists (its value is None)
print(list(listing.keys()))
````

### Sets

A set keeps each value once. It answers "which different values are there?" and "is this value among them?", and two sets can be combined.

````python
districts = ["Mitte", "Pankow", "Mitte", "Neukölln", "Pankow"]
unique = set(districts)
print(len(districts), len(unique))
print("Spandau" in unique)
print(unique & {"Mitte", "Spandau"})     # intersection: in both sets
````

### Collections inside collections

A table is a collection of records. A **list of dictionaries** stores one record per element; each dictionary has the same keys. This is how data often arrive from a web service, and the next block turns such a list into a pandas table.

````python
records = [
    {"district": "Pankow", "nightly_price": 75, "available": True},
    {"district": "Mitte", "nightly_price": 90, "available": False},
    {"district": "Neukölln", "nightly_price": 110, "available": True},
]
print(len(records))                    # number of rows
print(records[1]["district"])          # the district of the second row
````

Lists and dictionaries are **mutable**: their content can change after they were created. A name refers to an object; it is not a box that holds a copy. The exercises contain a short puzzle about what this means when two names refer to the same list.

## Getting help: `help()`, `dir()` and the documentation

Python has help built in, which is often faster than a search and always matches the installed version:

- `help(x)` shows the documentation of a function, method or type.
- `dir(x)` lists the names (attributes and methods) available on a value. Names with double underscores, such as `__len__`, are used by Python internally and can be ignored for now.
- In a notebook, `str.split?` in a code cell shows the same text as `help`; in VS Code, hovering over a name shows its documentation.
- At home: the official [Python documentation](https://docs.python.org/3/) and the documentation of each package (for example [pandas](https://pandas.pydata.org/docs/)) are the authoritative sources.

````python
help(str.split)
````

````python
print([name for name in dir(list) if not name.startswith("_")])
````

(The expression in brackets keeps only the names without a leading underscore; it is a *list comprehension*, taught in the next block.)

**Error messages are help, too.** When a cell fails, Python prints a **traceback**. Read it from the bottom: the last line names the type of error and the reason, the lines above show where it happened.

```text
TypeError                                 Traceback (most recent call last)
Cell In[12], line 1
----> 1 "80" + 20

TypeError: can only concatenate str (not "int") to str
```

| Error | Typical cause |
|---|---|
| `SyntaxError` | the code is not valid Python: a missing bracket, quote or colon |
| `NameError` | a name that was never assigned, often a typo or a cell that was not run |
| `TypeError` | an operation on the wrong type, such as text plus a number |
| `KeyError` | a key that the dictionary does not have |
| `IndexError` | a position beyond the end of a list |

> **In practice.** Before asking anyone (or any AI assistant), read the last line of the traceback, look at the values and types involved with `print()` and `type()`, and try the smallest example that shows the problem. Most errors are found this way.
